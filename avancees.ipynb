### On a regardé la matrice de colinéarité de nos covariables

![Matrice de correlation](./matrice_correlation.png).

On voit qu'on a une maxi corrélation entre les variables de fonte des glaces (logique) et qu'elles sont également très corrélées avec la température.
On a donc regroupé ces 2 variables en une seule qui est la moyenne des 2 : ice_mass

Ensuite on a calculé le VIF de chaque covariable : c'est une mesure du fait qu'une covariable est très bien expliquée par les autres covariables. Si grand VIF (>10),  c'est que la covariable est très bien expliquée par les autres covariables => colinéarité.

| Feature         | VIF       |
|-----------------|-----------|
| ice_mass        | 34.626983 |
| sea_temperature | 34.625976 |
| CO2             | 1.429604  |
| sea_salinity    | 1.109809  |
| chlorophylle    | 1.053530  |


comme ice_mass et sea_temperature on un grand VIF, il faut faire qqchose. Le plus simple c'est d'en virer une des 2 (sinon ACP mais on perd de l'interprétabilité). Selon la littérature on préfère virer la ice_mass

| Feature         | VIF       |
|-----------------|-----------|
| CO2             | 1.419966  |
| sea_temperature | 1.406537  |
| sea_salinity    | 1.060321  |
| chlorophylle    | 1.049594  |

Voila ce qu'on obtient après avoir enlevé la masse de glace. Maintenant on est bon, on peut faire une régression polynomiale sur ces 4 variables.
