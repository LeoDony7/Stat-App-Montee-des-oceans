```python
import xarray as xr
import pandas as pd
import glob

# 📂 Chemin vers le dossier contenant les fichiers NetCDF (dans ton dépôt Git)
chemin_fichiers = "BDD_ESA_SSS/*.nc"  

# 📌 Liste des fichiers NetCDF
fichiers_nc = glob.glob(chemin_fichiers)

# 🏗️ Initialiser une liste pour stocker les données
data_list = []

# 🔄 Boucle sur chaque fichier
for fichier in fichiers_nc:
    ds = xr.open_dataset(fichier)
    
    # 🌍 Moyenne spatiale sur toute la zone
    variable = "sss"  # ⚠️ Remplace par le vrai nom de la variable
    df = ds[variable].mean(dim=["lat", "lon"]).to_dataframe().reset_index()

    # 📌 Garder uniquement le temps et la moyenne
    df_final = df[["time", variable]].rename(columns={variable: "moyenne_variable"})

    # 🔄 Convertir la colonne 'time' en format datetime et appliquer le format 'YYYY-MM'
    df_final['time'] = pd.to_datetime(df_final['time']).dt.strftime('%Y-%m')

    # 📌 Stocker les résultats
    data_list.append(df_final)

# 🔗 Concaténer toutes les données
df_resultat = pd.concat(data_list)

# 💾 Sauvegarder le fichier CSV dans le même dépôt Git
df_resultat.to_csv("BDD_ESA_SSS/sortie.csv", index=False)

print("✅ Fichier 'sortie.csv' créé dans 'BDD_ESA_SSS' avec les colonnes 'time' (format YYYY-MM) et 'moyenne_variable'.")
```
